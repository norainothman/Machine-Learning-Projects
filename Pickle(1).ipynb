```python
# Step 3 : Data Preparation
# a.) check for missing and duplicate values
# b.) convert non-numerical features into numerical features
# c.) normalization
# d.) select independent and dependent variable
```

```python
# a.) check for missing and duplicate values
data.isnull().sum()
```

```python
data.duplicated().sum()
```

```python
# b.) convert non-numerical features into numerical features
import sklearn.preprocessing as pp
lb = pp.LabelBinarizer()
```

```python
# LabelBinarizer neg_label  --> int, default=0    pos_label --> int, default=1
print(lb.fit_transform(['fail','pass']))
print(lb.fit_transform(['true','false']))
print(lb.fit_transform(['yes','no']))
print(lb.fit_transform(['positive','negative']))
print(lb.fit_transform(['success','no success']))
```

```python
data.Status = lb.fit_transform(data.Status)
```

```python
data.info()
```

```python
# c.) normalization
# Normalization is a data preprocessing step in machine learning where we adjust
# the scale of numeric features so that they fall within a similar range (usually
# between 0 and 1, or –1 and 1).

# normalize these numbers for me
# numbers = [57,  66,  73,  76,  79,  81,  90,  96, 100]

# Xnorm = (X-Xmin) / (Xmax - Xmin)        known as Min-Max Scaling
# Xmin = 57
# Xmax = 100
# X = 96  and X = 100

# Xnorm = ??


# 57 --> 0
# 96 --> 0.9
# 100 --> 1
```

```python
vals = data.Test_Grade.values
print(vals)

normalized_values = pp.normalize([vals])
print(normalized_values)

min_max_sc = pp.MinMaxScaler()
normalized_values1 = min_max_sc.fit_transform([vals])
print(normalized_values1)
```

```python
# d.) select independent and dependent variable
x = data.Hours_Studied.values
x = x.reshape(9,1)
print(x)
```

```python
y = data.Test_Grade.values
print(y)
```

```python
# Step 4 : Model Development
# Apply A Suitable ML Algo On Dataset

import sklearn.linear_model as lm
lin_reg = lm.LinearRegression()
```

```python
lin_reg.fit(x,y)
```

```python
'''
Using Pickle

Using pickle in machine learning is mainly about saving and loading trained models in Python so you don’t have to retrain them every time.


Why use pickle in Machine Learning?

When you train a model (e.g. using scikit-learn), training can take time. Instead of retraining:

Save the model once
Load it later for predictions'''

print('Using Pickle')

import pickle
# dump() is used to save ML Model
# pickle uses bytestream format
pickle.dump(lin_reg,open('slrname.pkl','wb')) # w --> write / create a file  b --> bytestream
```

```python
# Step 5 : Evaluate Our Model Performance
data['predictedValues'] = lin_reg.predict(x)
```
